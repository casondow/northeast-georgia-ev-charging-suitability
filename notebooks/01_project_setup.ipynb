# EV Charging Infrastructure Suitability and Demand Analysis

## Study Area

This project evaluates public electric vehicle charging infrastructure across Gwinnett, Barrow, Hall, Jackson, Clarke, Oconee, and Walton counties in Northeast Georgia.

## Research Question

Where should additional public EV charging stations be installed, what charging speed and connector type should each location use, and how many charging ports are needed?

## Project Objectives

1. Map existing public EV charging stations and charging ports.
2. Classify stations by charging level, power, and connector type.
3. Identify communities and transportation corridors with inadequate charging access.
4. Estimate charging-port demand using population, household, income, housing, and transportation variables.
5. Identify suitable locations for Level 2 and DC fast-charging infrastructure.
6. Recommend charging technology, connector types, and port counts for candidate sites.
